### import required packages

```python
import nltk

nltk.download('punkt_tab')
nltk.download('wordnet')
nltk.download('omw-1.4')
nltk.download('stopwords')
```

### load the data

```python
data = """Police have fired tear gas and attacked supporters of the youth-led Cockroach movement with canes and baton charges as more than 10,000 marched towards India’s parliament in New Delhi in the largest protest yet demanding the resignation of the education minister and a crackdown on corruption in the exam system. The march on Monday by the protesters who ironically call themselves the Cockroach Janta Party to mock the name of the ruling party of Prime Minister Narendra Modi was even bigger than expected, with numbers swelled by anger at Saturday’s forcible hospitalisation of veteran 59-year-old activist Sonam Wangchuk, who has been on hunger strike for three weeks in solidarity with the movement. As the protesters marched, they chanted slogans calling for Modi and Education Minister Dharmendra Pradhan to quit. Modi’s government, shaken by the emergence of the Cockroach movement in May, offered a minor concession by inviting two activists to meet Health Minister JP Nadda for talks."""
```

### sentence segmentation

```python
# splitting the data into individual sentences
sentences = nltk.tokenize.sent_tokenize(data)
print(sentences)
```

```python
print(f"No Of Sentences : {len(sentences)}")
```

### word tokenization

```python
# collect all the words
all_words = []

for sentence in sentences:
    # split every sentence into individual words
    words = nltk.tokenize.word_tokenize(sentence)

    # append the found words to the all words collection
    all_words.extend(words)

# print all the words
print(all_words)
```

```python
# collect all the words
all_words = []

# split the data into individual words
all_words = nltk.tokenize.word_tokenize(data)

# print all the words
print(all_words)
print(f"no of words = {len(all_words)}")
```

### POS Tagging

```python
from nltk.tag import pos_tag

# get the pos tagging of all the words
pos_tags=pos_tag(all_words)
```

### Stemminng

```python
from nltk.stem import PorterStemmer

# create a stemmer object
stemmer = PorterStemmer()

# example of words
words = ['better', 'running', 'went', 'gone', 'apples', 'ran', 'best']

# reduce all the words to their base forms
for word in words:
    base_form = stemmer.stem(word)
    print(f"{word} = {base_form}")
```

### lemmatization

```python
from nltk.stem import WordNetLemmatizer

# create a lemmatizer
lemmatizer = WordNetLemmatizer()

# example of words
words = [
    ('better', 'a'),
    ('running', 'v'),
    ('went', 'v'),
    ('gone', 'v'),
    ('apples', 'n'),
    ('ran', 'v'),
    ('better', 'a')
]

# reduce all the words to their base forms
for word, pos_tag in words:
    base_form = lemmatizer.lemmatize(word, pos_tag)
    print(f"{word} = {base_form}")
```

```python
# reduce all the words to their base forms
for word, pos_tag in pos_tags:
    tag = 'n'
    if pos_tag.startswith('N'):
        tag = 'n'
    elif pos_tag.startswith('V'):
        tag = 'v'
    elif pos_tag.startswith('J'):
        tag = 'v'
    elif pos_tag.startswith('A'):
        tag = 'a'

    base_form = lemmatizer.lemmatize(word, tag)
    print(f"{word} {pos_tag} [{tag}] = {base_form}")
```

### lowercasing

```python
lowercased_words = [word.lower() for word in all_words]
print(lowercased_words)
print(f"total no of words = {len(lowercased_words)}")
```

### removing special symbols

```python
symbols = ['.', ',', '"', "'", ';', ':', '’']

# remove the symbols
all_words_without_symbols = [word for word in lowercased_words if word not in symbols]
print(all_words_without_symbols)
print(f"total no of words without symbols = {len(all_words_without_symbols)}")
```

### remove the stop words

```python
from nltk.corpus import stopwords

# find the stopword from english language
stop_words = stopwords.words('english')
print(f"stop words = {stop_words}")
```

```python
# remove the stop words
all_words_without_stopwords = [word for word in all_words_without_symbols if word not in stop_words]
print(all_words_without_stopwords)
print(f"total no of words without stopwords = {len(all_words_without_stopwords)}")
```

# RNN Demo

### import required packages

```python
import numpy
import torch
```

### get the data

```python
# create a dummy data
x = torch.tensor([
    [
        [10.0, 20.0, 30.0, 40.0, 50.0],
        [10.0, 20.0, 30.0, 40.0, 30.0],
        [10.0, 20.0, 30.0, 70.0, 50.0],
        [10.0, 20.0, 90.0, 40.0, 50.0],
        [10.0, 10.0, 30.0, 40.0, 50.0],
    ]
])
```

### create the RNN layer

```python
# create rnn
rnn = torch.nn.RNN(
    input_size=5,
    hidden_size=5,
    batch_first=True
)
```

```python
# get the shape of x
x.shape
```

```python
# pass the data to rnn
output, hidden_state = rnn(x)
```

```python
# get the output
print(output)
```

```python
# get the hidden state
print(hidden_state)
```

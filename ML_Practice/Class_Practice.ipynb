## company gives you a 1D array of 48 values:

```python
import numpy as np

sales = np.array([
    45, 52, 60, np.nan, 70, 48, 55, 62, 58, 49, 65, 72,
    50, np.nan, 63, 68, 75, 45, 52, 60, 55, 48, 67, 74,
    42, 48, np.nan, 65, 72, 40, 50, 58, 54, 46, 62, 70,
    47, 53, 61, 66, np.nan, 44, 51, 59, 56, 47, 64, 71
])
print(sales)
```

## Reshape into (4,12)

```python
sales_reshape = sales.reshape(4, 12)
print(sales_reshape)
```

## Replace NaN with Weekly (Column) Mean

```python
filled_sales = np.where(
    np.isnan(sales_reshape),
    np.nanmean(sales_reshape , axis=0),
    sales_reshape
)
print(filled_sales)
```

## Find Top-Performing Store

```python
store_totals = np.sum(filled_sales, axis=1)
print(store_totals)
```

## Store with highest sales

```python
top_store = np.argmax(store_totals)
print(top_store)
```

## Flag Weeks Where ALL Stores Sold Below ₹50k

```python
low_sales_weeks = np.all(filled_sales < 50, axis=0)

print(low_sales_weeks)
```

```python
# Q43.
# From arr = np.arange(16).reshape(4, 4):
# 1. Split into 2 equal parts along axis=0 using np.vsplit
# 2. Split into 4 equal parts along axis=1 using np.hsplit
# 3. Reassemble the vsplit result using np.vstack and verify equality

import numpy as np

arr = np.arange(16).reshape(4, 4)

print(f"Original Array:\n{arr}")

# Split into 2 equal parts vertically (rows)
v_parts = np.vsplit(arr, 2)

print("\nVertical Split:")
for i, part in enumerate(v_parts):
    print(f"\nPart {i+1}:\n{part}")

# Split into 4 equal parts horizontally (columns)
h_parts = np.hsplit(arr, 4)

print("\nHorizontal Split:")
for i, part in enumerate(h_parts):
    print(f"\nPart {i+1}:\n{part}")

# Reassemble using vstack
reassembled = np.vstack(v_parts)

print(f"\nReassembled Array:\n{reassembled}")

# Verify equality
print(f"\nArrays Equal? {np.array_equal(arr, reassembled)}")
```

```python
# Q81.
# Compute row-wise L2 (Euclidean) distance

import numpy as np

# Example arrays
A = np.random.rand(5, 4)
B = np.random.rand(5, 4)

print("A:")
print(A)

print("\nB:")
print(B)

# Row-wise L2 distance
distances = np.sqrt(np.sum((A - B) ** 2, axis=1))

print("\nRow-wise L2 Distances:")
print(distances)

# Closest pair of rows
closest_idx = np.argmin(distances)

print("\nIndex of Closest Row Pair:")
print(closest_idx)

print("\nMinimum Distance:")
print(distances[closest_idx])
```

```python
# Q82.
# Find the value closest to the mean of each column

import numpy as np

# 100 samples, 5 features
X = np.random.rand(100, 5)

print("Shape of X:", X.shape)

# Column means
col_means = X.mean(axis=0)

# Index of value closest to each column mean
indices = np.argmin(
    np.abs(X - col_means),
    axis=0
)

# Values at those positions
values = X[indices, np.arange(X.shape[1])]

print("\nColumn Means:")
print(col_means)

print("\nIndices Closest to Column Means:")
print(indices)

print("\nValues Closest to Column Means:")
print(values)
```

```python
# Q83.
# Compare float64 vs float32 memory and computation speed

import numpy as np
import time

# float64 array (default)
arr64 = np.random.rand(1000, 1000)

# float32 version
arr32 = arr64.astype(np.float32)


# Time float64 sum
start = time.perf_counter()
sum64 = arr64.sum()
time64 = time.perf_counter() - start


# Time float32 sum
start = time.perf_counter()
sum32 = arr32.sum()
time32 = time.perf_counter() - start

# Means
mean64 = arr64.mean()
mean32 = arr32.mean()

# Memory usage
bytes64 = arr64.nbytes
bytes32 = arr32.nbytes

print("float64")
print("dtype:", arr64.dtype)
print("nbytes:", bytes64)
print("mean:", mean64)
print("sum time:", time64)

print("\nfloat32")
print("dtype:", arr32.dtype)
print("nbytes:", bytes32)
print("mean:", mean32)
print("sum time:", time32)

print("\nMemory Ratio (64/32):")
print(bytes64 / bytes32)
```

```python
# Q84.
# Demonstrate the difference between ravel() and flatten()

import numpy as np

arr = np.arange(9).reshape(3, 3)

print("Original Array:")
print(arr)


# ravel() -> view
r = arr.ravel()

r += 1

print("\nAfter ravel() += 1:")
print(arr)


# flatten() -> copy
arr2 = np.arange(9).reshape(3, 3)

f = arr2.flatten()

f += 1

print("\nAfter flatten() += 1:")
print("Flattened Copy:")
print(f)

print("\nOriginal Array Remains Unchanged:")
print(arr2)
```

```python
# Q85.
# Remove rows containing outliers using the IQR rule

import numpy as np

# Random dataset: 200 samples, 4 features
data = np.random.rand(200, 4)

print("Original Shape:", data.shape)

# Q1 and Q3 for each column
Q1 = np.percentile(data, 25, axis=0)
Q3 = np.percentile(data, 75, axis=0)

# IQR
IQR = Q3 - Q1

# Lower and upper bounds
lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

print("\nQ1:")
print(Q1)

print("\nQ3:")
print(Q3)

print("\nLower Bounds:")
print(lower)

print("\nUpper Bounds:")
print(upper)

# Mask: each value must be within its column bounds
mask = (data >= lower) & (data <= upper)

# Keep rows where ALL 4 features satisfy the condition
row_mask = np.all(mask, axis=1)

filtered_data = data[row_mask]

print("\nRows Remaining:", filtered_data.shape[0])
print("Filtered Shape:", filtered_data.shape)
```

```python
"""Scenario
A retail company has recorded weekly sales (in ₹ thousands) for 4 stores over 12 weeks. The data arrived as a flat 1D array of 48 values with some entries corrupted as np.nan. Your task is to restructure and analyze it.
Tasks"""
import numpy as np
a = np.random.randint(1000, 2000, 48).astype(np.float64)
a = a.reshape(4, 12)
a[0,2] = np.nan
a[0,7] = np.nan
a[1,3] = np.nan
a[2,5] = np.nan
a[3,1] = np.nan
a[3,9] = np.nan
print(f"reshaped array : \n{a}")

col_means = np.nanmean(a, axis=0)
print(f"columnwise mean: \n{col_means}")

nan_rows, nan_cols = np.where(np.isnan(a))
print(f"nan_row : {nan_rows}\nnan_col : {nan_cols}")
a[nan_rows, nan_cols] = col_means[nan_cols]


top_store = np.argmax(a.sum(axis=1))
print("Top store:", top_store + 1)


low_weeks = np.all(a < 50, axis=0)
print("Low weeks mask:", low_weeks)

```

## 🔹 Section 1: Basics & Series (Q1–Q20)

### Q1. Create a Pandas Series from a list of integers.
> 💡 **Hint:** Use `pd.Series([...])`. The default index starts from 0.

```python
import pandas as pd
l=pd.Series([10,20,30,40,50])
print(l)
print(type(l))
```

### Q2. Create a Series from a dictionary with custom index.
> 💡 **Hint:** Pass a `dict` to `pd.Series()`. Keys become the index labels automatically.

```python
d={
    "Name": 'Vikas',
    "Marks":90,
    "Age":25
}
dic=pd.Series(d)
print(dic)
```

### Q3. Access elements using index labels.
> 💡 **Hint:** Use `s["label"]` for a single value, `s[["a","b"]]` for multiple, and `s["a":"c"]` for a range. Label slices are **inclusive** on both ends.

```python
l=pd.Series(
    [1,2,3,4,5],
    index=['a','b','c','d','e']
)
print(l)
print(l["a"])
print(l[['b','c']])
print(l["a":"e"])
```

### Q4. Perform arithmetic operations on a Series.
> 💡 **Hint:** Use `+`, `-`, `*`, `/`, `//`, `%`, `**` directly on the Series. Operations are **element-wise** and **index-aligned**.

```python
import pandas as pd
l=pd.Series([10,20,30,40,50])
print(l)
print("*"*50)
print("Addition :\n", l + 5)
print("*"*50)
print("Subtraction :\n", l - 5)
print("*"*50)
print("Multiplication :\n", l * 5)
print("*"*50)
print("Division :\n", l / 5)
print("*"*50)
print("Floor Division : \n ", l // 5 )
print("*"*50)
print("Modulus : \n ", l % 5)
print("*"*50)
print("Power : \n ", l ** 2)
```

### Q5. Find mean, min, max of a Series.
> 💡 **Hint:** Use `.mean()`, `.min()`, `.max()`, `.sum()`, `.median()`, `.std()` directly on the Series.

```python
l=pd.Series([10,20,30,40,50])
print(l)
print("Mean:", l.mean())
print("Min :", l.min())
print("Max :", l.max())
print("Sum :", l.sum())
print("Median :", l.median())
print("Std :", l.std())
```

### Q6. Apply conditional filtering on a Series.
> 💡 **Hint:** Use a boolean mask: `s[s > 50]`. You can also chain conditions with `&` and `|`.

```python
import pandas as pd
l = pd.Series([25, 60, 45, 80, 90])
print(l[l > 50])
print(l[(l > 40) & (l < 90)])
print(l[(l < 30) | (l > 80)])
```

### Q7. Replace values in a Series.
> 💡 **Hint:** Use `.replace(old_val, new_val)` or pass a dictionary `{old: new}` to replace multiple values at once.

```python
l=pd.Series([10,20,30,40,50])
l_rep=l.replace(20,99)
print(l_rep)
```

### Q8. Convert Series data type.
> 💡 **Hint:** Use `.astype(dtype)` — e.g., `.astype(int)`, `.astype(float)`, `.astype("category")`.

```python
s=pd.Series([10,20,30])
s_f=s.astype(float)
print(s_f)
print(s_f.dtype)
```

### Q9. Check for null values in Series.
> 💡 **Hint:** Use `.isnull()` (returns boolean Series) or `.isnull().sum()` for a count.

```python
import pandas as pd
import numpy as np
s = pd.Series([10, np.nan, 30, np.nan, 50])
print(s.isnull())
```

### Q10. Fill missing values in Series.
> 💡 **Hint:** Use `.fillna(value)` for constant, `.fillna(s.mean())` for mean, `.ffill()` for forward fill, `.bfill()` for backward fill.

```python
import pandas as pd
import numpy as np
s=pd.Series([10,np.nan,30,np.nan,50])
s_fill=s.fillna(0)
print(s_fill)
print("*"*50)
mean_val=s.mean()
s_filled=s.fillna(mean_val)
print(s_filled)
print("*"*50)
print(s.ffill())
```

## 🔹 Section 2: Indexing, Selection & Slicing (Q21–Q40)

### Q21. Select rows using `loc`.
> 💡 **Hint:** `df.loc[label]` for a single row, `df.loc[label1:label2]` for a range. **Inclusive on both ends.**

```python
import pandas as pd
df = pd.DataFrame({
    "Name": ["Vikas", "Rahul", "Amit", "Sneha"],
    "Marks": [90, 85, 78, 95],
    "Age": [25, 24, 23, 22]
}, index=["A", "B", "C", "D"])

print(df)
print("*"*50)
#select single row
print(df.loc["A"])
print("*"*50)
#select multiple rows using range
print(df.loc["A":"C"])
```

### Q22. Select rows using `iloc`.
> 💡 **Hint:** `df.iloc[0]` for first row, `df.iloc[1:4]` for rows 1–3. **Exclusive on the end** (Python-style).

```python
import pandas as pd

df = pd.DataFrame({
    "Name": ["Vikas", "Rahul", "Amit", "Sneha"],
    "Marks": [90, 85, 78, 95],
    "Age": [25, 24, 23, 22]
})

print(df)
print("*"*50)
#select single row
print(df.iloc[0])
print("*"*50)
#select third row
print(df.iloc[2])
print("*"*50)
#select multiple rows using slicing
print(df.iloc[1:4])
```

### Q23. Access specific row and column.
> 💡 **Hint:** `df.loc[row_label, "col_name"]` or `df.iloc[row_pos, col_pos]` for a single cell.

```python
df = pd.DataFrame({
    "Name": ["Vikas", "Rahul", "Amit"],
    "Marks": [90, 85, 78],
    "Age": [25, 24, 23]
}, index=["A", "B", "C"])

print(df)
print(df.loc["B", "Marks"])
print(df.loc["C", "Age"])
print(df.iloc[1, 1])
```

### Q24. Select subset using multiple rows & columns.
> 💡 **Hint:** `df.loc[[r1,r2], ["col1","col2"]]` or `df.iloc[[0,1], [0,2]]`.

```python
df = pd.DataFrame({
    "Name": ["Vikas", "Rahul", "Amit", "Sneha"],
    "Marks": [90, 85, 78, 95],
    "Age": [25, 24, 23, 22]
}, index=["A", "B", "C", "D"])

print(df)
print(df.loc[["A", "C"], ["Name", "Marks"]])
print(df.loc[["B", "D"], ["Name"]])
```

### Q25. Apply conditional filtering using `loc`.
> 💡 **Hint:** `df.loc[df["col"] > value, ["col1","col2"]]` — pass a boolean Series as the row selector.

```python
import pandas as pd

df = pd.DataFrame({
    "Name": ["Vikas", "Rahul", "Amit", "Sneha"],
    "Marks": [90, 85, 78, 95],
    "Age": [25, 24, 23, 22]
})

print(df)
print(df.loc[df["Marks"] > 80])
df.loc[df["Marks"] > 80]
print(df.loc[df["Marks"] > 80, ["Name", "Marks"]])
print(df.loc[df["Age"] > 23])
print(df.loc[(df["Marks"] > 80) & (df["Age"] > 23)])
print(df.loc[(df["Marks"] > 90) | (df["Age"] < 23)])
```

### Q26. Slice rows using index range.
> 💡 **Hint:** Use `df.iloc[start:end]` for position-based slicing or `df.loc[label1:label2]` for label-based slicing.

```python
import pandas as pd
df = pd.DataFrame({
    "Name": ["Vikas", "Rahul", "Amit", "Sneha", "Neha"],
    "Marks": [90, 85, 78, 95, 88],
    "Age": [25, 24, 23, 22, 21]
}, index=["A", "B", "C", "D", "E"])

print(df)
print("*"*50)
#position based slicing using iloc
print(df.iloc[1:4])
print("*"*50)
#first 3 rows
print(df.iloc[0:3])
print("*"*50)
#labeled base slicing using loc
print(df.loc["B":"D"])
```

### Q27. Slice columns using labels.
> 💡 **Hint:** `df.loc[:, "col1":"col3"]` — selects all rows, columns from col1 to col3 (label-inclusive).

```python
df = pd.DataFrame({
    "Name": ["Vikas", "Rahul", "Amit"],
    "Marks": [90, 85, 78],
    "Age": [25, 24, 23],
    "City": ["Mumbai", "Pune", "Delhi"]
})

print(df)
print("*"*50)
#select columns name to age
print(df.loc[:, "Name":"Age"])
print("*"*50)
#select marks to city
print(df.loc[:, "Marks":"City"])
print("*"*50)
#select all columns
print(df.loc[:, :])
```

### Q28. Slice columns using position.
> 💡 **Hint:** `df.iloc[:, 1:4]` — selects all rows, columns at positions 1, 2, 3.

```python
#select columns from position 1 to 2
print(df.iloc[:, 1:3])
print("*"*50)
#select first two columns
print(df.iloc[:, 0:2])
```

### Q29. Perform combined row & column slicing.
> 💡 **Hint:** `df.loc[row_condition, ["col1","col2"]]` or `df.iloc[0:5, [0, 2, 4]]`.

```python
df = pd.DataFrame({
    "Name": ["Vikas", "Rahul", "Amit", "Sneha", "Neha"],
    "Marks": [90, 85, 78, 95, 88],
    "Age": [25, 24, 23, 22, 21],
    "City": ["Mumbai", "Pune", "Delhi", "Chennai", "Bangalore"]
})

print(df)
print("*"*50)
#using loc
print(df.iloc[0:3, 0:2])
print("*"*50)
#select specfic rows and columns
print(df.iloc[[0, 3, 4], [0, 2]])
```

### Q30. Set a column as index.

### Q31. Create hierarchical index (MultiIndex).
> 💡 **Hint:** `df.set_index(["col1","col2"])` — pass a **list** of column names to create a MultiIndex.

```python
df = pd.DataFrame({
    "Department": ["IT", "IT", "HR", "HR"],
    "Name": ["Vikas", "Rahul", "Amit", "Sneha"],
    "Salary": [50000, 60000, 55000, 58000]
})

print(df)
print("*"*50)
#Creating multiindex
df2 = df.set_index(["Department", "Name"])
print(df2)
print("*"*50)
#accessing data using multiindex
print(df2.loc[("IT", "Vikas")])
```

### Q32. Reset index.
> 💡 **Hint:** `df.reset_index()` — moves the current index back to a column. Use `inplace=True` to modify in-place.

```python
import pandas as pd
df = pd.DataFrame({
    "Name": ["Vikas", "Rahul", "Amit"],
    "Marks": [90, 85, 78]
})
df = df.set_index("Name")
print(df)
print("*"*50)
df.reset_index(inplace=True)
print(df)
```

### Q33. Reset multi-index.
> 💡 **Hint:** `df.reset_index()` on a MultiIndex DataFrame flattens both index levels back to columns.

```python
df = pd.DataFrame({
    "Department": ["IT", "IT", "HR", "HR"],
    "Name": ["Vikas", "Rahul", "Amit", "Sneha"],
    "Salary": [50000, 60000, 55000, 58000]
})
df = df.set_index(["Department", "Name"])
print(df)
print("*"*50)
df2 = df.reset_index()
print(df2)
```
